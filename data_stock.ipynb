```python
import pandas as pd 
import numpy as np
import streamlit as st
import matplotlib.pyplot as plt
import os

```

```python
df["date"] = pd.to_datetime(df["date"])
df = df.sort_values(["Ticker", "date"])
```

```python
numeric_cols = ['open', 'high', 'low', 'close', 'volume']
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric)
```

```python
df["Daily_Return"] = df.groupby("Ticker")["close"].pct_change()
df["Daily_Return"]
```

```python
df["Cumulative_Return"] = (1 + df["Daily_Return"]).groupby(df["Ticker"]).cumprod()
df["Cumulative_Return"]
```

```python
df["Month"] = df["date"].dt.to_period("M")
```

```python
df["Month"]
```

```python
sector_df = pd.read_csv("C:/Users/arunm/Downloads/Sector_data - Sheet1.csv")
```

```python
sector_df.head()
```

```python
sector_df["Symbol"] = sector_df["Symbol"].str.split(":").str[1].str.strip()
```

```python
sector_df["Symbol"]
```

```python
df = df.merge(
        sector_df[["Symbol", "sector"]],
        left_on="Ticker",
        right_on="Symbol",
        how="left"
)

df.drop(columns="Symbol", inplace=True)
```

```python
yearly_return = (
    df.groupby("Ticker")["Cumulative_Return"]
      .last()
      .reset_index()
      .rename(columns={"Cumulative_Return":"Yearly_Return"})
)
```

```python
yearly_return.head()
```

```python
os.makedirs("outputs", exist_ok=True)

top_green = yearly_return.sort_values("Yearly_Return", ascending=False).head(10)
top_green.to_csv("top10_green.csv", index=False)
```

```python
top_green
```

```python
top_red = yearly_return.sort_values("Yearly_Return").head(10)
top_red.to_csv("outputs/top10_red.csv", index=False)
```

```python
top_red
```

```python
total = yearly_return.shape[0]

green = (yearly_return["Yearly_Return"] > 1).sum()
red   = (yearly_return["Yearly_Return"] < 1).sum()

market_summary = {
    "Total Stocks": total,
    "Green Stocks": green,
    "Red Stocks": red,
    "Green (%)": round(green/total*100,2)
}

market_summary
```

```python
avg_price = df.groupby("Ticker")["close"].mean().mean()

avg_volume = df.groupby("Ticker")["volume"].mean().mean()

print("Avg Price:", avg_price)
print("Avg Volume:", avg_volume)
```

```python
vol = (
    df.groupby("Ticker")["Daily_Return"]
      .std()
      .reset_index()
      .rename(columns={"Daily_Return":"Volatility"})
      .sort_values("Volatility", ascending=False)
)

vol.to_csv("outputs/volatility.csv", index=False)
vol.head()
```

```python
cum = df.groupby(["Ticker","date"])["Cumulative_Return"].last().reset_index()
cum.to_csv("outputs/cumulative_returns.csv", index=False)
cum.head()
```

```python
sector_perf = df.groupby("sector")["Cumulative_Return"].last()

sector_perf = sector_perf.reset_index()
sector_perf.columns = ["Sector", "Avg_Yearly_Return"]

sector_perf.to_csv("outputs/sector_performance.csv", index=False)
sector_perf.head()
```

```python
pivot = df.pivot(index="date", columns="Ticker", values="close")

corr = pivot.corr()

corr.to_csv("outputs/correlation_matrix.csv")
corr.head()
```

```python
monthly = (
    df.groupby(["Ticker","Month"])["Cumulative_Return"]
      .last()
      .groupby("Month")
      .apply(lambda x: x.sort_values(ascending=False))
)

for m in monthly.index.levels[0]:
    month_data = monthly.loc[m]
    winners = month_data.head(5)
    losers  = month_data.tail(5)
    print(f"\nMONTH: {m}")
    print("TOP 5 GAINERS")
    print(winners)
    print("TOP 5 LOSERS")
    print(losers)
```
